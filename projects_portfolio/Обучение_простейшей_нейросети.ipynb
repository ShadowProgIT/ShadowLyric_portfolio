```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

X = np.array([[0, 0],
              [0, 1],
              [1, 0],
              [1, 1]], dtype=float)

y = np.array([[0],
              [1],
              [1],
              [0]], dtype=float)

W1 = np.random.randn(2, 2) * 0.2  
b1 = np.zeros((1, 2))             

W2 = np.random.randn(2, 1) * 0.2  
b2 = np.zeros((1, 1))             

def sigmoid(z):
    z = np.clip(z, -500, 500)
    return 1 / (1 + np.exp(-z))

learning_rate = 0.5    
epochs = 30000         

loss_history = []

for epoch in range(epochs):
    z1 = X @ W1 + b1
    a1 = sigmoid(z1)

    z2 = a1 @ W2 + b2
    a2 = sigmoid(z2)

    loss = np.mean((a2 - y) ** 2)
    loss_history.append(loss)  

    if epoch % 4000 == 0:
        print(f"Эпоха {epoch:5d} | Потери (MSE): {loss:.6f}")

    dL_da2 = a2 - y
    da2_dz2 = a2 * (1 - a2)      
    delta2 = dL_da2 * da2_dz2    

    dL_dW2 = a1.T @ delta2
    dL_db2 = np.sum(delta2, axis=0, keepdims=True)

    delta1 = (delta2 @ W2.T) * (a1 * (1 - a1))

    dL_dW1 = X.T @ delta1
    dL_db1 = np.sum(delta1, axis=0, keepdims=True)

    W2 -= learning_rate * dL_dW2
    b2 -= learning_rate * dL_db2
    W1 -= learning_rate * dL_dW1
    b1 -= learning_rate * dL_db1

z1_final = X @ W1 + b1
a1_final = sigmoid(z1_final)
z2_final = a1_final @ W2 + b2
predictions = sigmoid(z2_final)

print("\nРезультаты обучения")
print("Предсказания (до округления):")
print(predictions)
print("\nОкруглённые предсказания:")
print(np.round(predictions).astype(int))
print("\nОжидаемые значения:")
print(y.astype(int))

plt.figure(figsize=(10, 6))
plt.plot(loss_history, color='darkblue', linewidth=1)
plt.title('Динамика функции потерь', fontsize=14)
plt.xlabel('Эпоха', fontsize=12)
plt.ylabel('MSE', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
